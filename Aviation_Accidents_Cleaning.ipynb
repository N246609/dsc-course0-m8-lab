```python
# List the injury columns
injury_columns = [
    "Total.Fatal.Injuries",
    "Total.Serious.Injuries",
    "Total.Minor.Injuries",
    "Total.Uninjured"
]

# Convert the injury columns to numbers
for col in injury_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Calculate the estimated total number of occupants
df["Total.Occupants"] = df[injury_columns].sum(
    axis=1,
    min_count=4
)

# Check that the new column exists
print(df[["Total.Occupants"]].head())
```

```python

# Combine fatal and serious injury counts.


df["Fatal.Serious.Injuries"] = df[
    ["Total.Fatal.Injuries", "Total.Serious.Injuries"]
].sum(axis=1, min_count=2)

# Calculate the proportion of occupants with fatal or serious injuries.


df["Fatal.Serious.Injury.Rate"] = (
    df["Fatal.Serious.Injuries"] / df["Total.Occupants"]
).where(df["Total.Occupants"] > 0)


df[
    [
        "Total.Occupants",
        "Fatal.Serious.Injuries",
        "Fatal.Serious.Injury.Rate"
    ]
].head(10)
```

```python

# Missing values remain missing.

df["Aircraft.Damage"] = (
    df["Aircraft.Damage"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Check the cleaned categories
print(df["Aircraft.Damage"].value_counts(dropna=False))
```

```python

# Recognize the damage categories 


known_damage = ["Destroyed", "Substantial", "Minor"]



df["Aircraft.Destroyed"] = np.nan

# Record confirmed destruction
df.loc[
    df["Aircraft.Damage"] == "Destroyed",
    "Aircraft.Destroyed"
] = 1


df.loc[
    df["Aircraft.Damage"].isin(["Substantial", "Minor"]),
    "Aircraft.Destroyed"
] = 0


print(df["Aircraft.Destroyed"].value_counts(dropna=False))
```

## Investigate the Make column

```python

print("Missing Make values:", df["Make"].isna().sum())

print("Unique Makes before cleaning:", df["Make"].nunique())


print(df["Make"].value_counts(dropna=False).head(20))
```

```python


print("Missing Make values:", df["Make"].isna().sum())


print("Unique Makes before cleaning:", df["Make"].nunique())


print(df["Make"].value_counts(dropna=False).head(20))
```

```python


df_clean = df.copy()


df_clean["Make"] = (
    df_clean["Make"]
    .astype("string")
    .str.strip()
    .str.upper()
)


df_clean["Make"] = df_clean["Make"].replace("", pd.NA)


df_clean = df_clean.dropna(subset=["Make"]).copy()


make_counts = df_clean["Make"].value_counts()


valid_makes = make_counts[make_counts >= 50].index

df_clean = df_clean[
    df_clean["Make"].isin(valid_makes)
].copy()

print("Rows remaining:", len(df_clean))
print("Manufacturers remaining:", df_clean["Make"].nunique())
print("\nTop 20 manufacturers:")
print(df_clean["Make"].value_counts().head(20))
```

## Cleaning other columns

```python

columns = [
    "Engine.Type",
    "Weather.Condition",
    "Number.of.Engines",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]

for col in columns:
    print("\n", col)
    print(df_clean[col].value_counts(dropna=False).head(15))
```

```python

text_columns = [
    "Engine.Type",
    "Weather.Condition",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]

for col in text_columns:
    df_clean[col] = (
        df_clean[col]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    
    df_clean[col] = df_clean[col].replace(
        ["", "UNKNOWN", "UNK", "N/A", "NONE", "NAN"],
        pd.NA
    )


df_clean["Number.of.Engines"] = pd.to_numeric(
    df_clean["Number.of.Engines"],
    errors="coerce"
)


invalid_engines = (
    (df_clean["Number.of.Engines"] <= 0) |
    (df_clean["Number.of.Engines"] % 1 != 0)
)

df_clean.loc[invalid_engines, "Number.of.Engines"] = pd.NA

print("Cleaning complete!")
```
